```python
def extract_salary_from(salary):
    if pd.isna(salary):
        return None

    text = str(salary).strip().lower()
    if text.startswith("от"):
        match = re.search(r"\d[\d \u00a0]*(?:[.,]\d+)?", text[2:])
    elif re.match(r"^\d", text) and "до" not in text:
        match = re.match(r"\d[\d \u00a0]*(?:[.,]\d+)?", text)
    else:
        return None

    if not match:
        return None

    amount = match.group().replace(" ", "").replace("\u00a0", "").replace(",", ".")
    return float(amount) if "." in amount else int(amount)
```

```python
def extract_salary_to(salary):
    if pd.isna(salary):
        return None

    text = str(salary).strip().lower()

    # Если в тексте есть диапазон "от ... до ..."
    if "до" in text:
        match = re.search(r"до\s*(\d[\d\s\xa0]*(?:[.,]\d+)?)", text)
    # Если это одно число, например "93 000 ₽"
    elif re.match(r"^\d", text):
        match = re.match(r"\d[\d\s\xa0]*(?:[.,]\d+)?", text)
    else:
        return None

    if not match:
        return None

    amount = match.group(1) if "до" in text else match.group()
    amount = amount.replace(" ", "").replace("\xa0", "").replace(",", ".")

    return float(amount) if "." in amount else int(amount)
```

```python
def extract_salary_currency(salary):
    if pd.isna(salary):
        return None

    match = re.search(
        r"₽|руб(?:\.|лей|ля)?|€|eur|евро|\$|usd|доллар(?:а|ов)?|₸|kzt|тенге",
        str(salary),
        flags=re.IGNORECASE,
    )

    if not match:
        return None

    currency = match.group().lower()

    if currency == "₽" or currency.startswith("руб"):
        return "RUB"
    if currency in {"€", "eur", "евро"}:
        return "EUR"
    if currency in {"$", "usd"} or currency.startswith("доллар"):
        return "USD"
    if currency in {"₸", "kzt", "тенге"}:
        return "KZT"

    return None
```

```python
def _get_experience_numbers(experience):
    if pd.isna(experience):
        return None

    text = str(experience).lower()

    if "нет опыта" in text:
        return [0, 0]

    numbers = re.findall(r"\d+", text)
    if not numbers:
        return None

    return [int(number) for number in numbers]


def experience_from(experience):
    numbers = _get_experience_numbers(experience)
    return numbers[0] if numbers else None


def experience_to(experience):
    numbers = _get_experience_numbers(experience)
    if not numbers:
        return None

    if "более" in str(experience).lower():
        return None

    return numbers[1] if len(numbers) > 1 else numbers[0]
```

```python
def get_superjob_salary_numbers(salary):
    if pd.isna(salary):
        return []

    matches = re.findall(r"(?<!\d)\d(?:[\d \u00a0]*\d)?(?:[.,]\d+)?", str(salary))
    numbers = []

    for match in matches:
        value = match.replace(" ", "").replace("\u00a0", "").replace(",", ".")
        numbers.append(float(value) if "." in value else int(value))

    return numbers


def superjob_salary_from(salary):
    numbers = get_superjob_salary_numbers(salary)
    return numbers[0] if numbers else None


def superjob_salary_to(salary):
    numbers = get_superjob_salary_numbers(salary)

    if not numbers:
        return None

    text = str(salary).lower()

    if len(numbers) > 1:
        return numbers[-1]
    if "от" in text:
        return None

    return numbers[0]


def superjob_experience_bounds(metadata):
    if pd.isna(metadata):
        return None, None

    text = str(metadata).lower()
    experience_match = re.search(r"опыт\s+работы", text)

    if not experience_match:
        return None, None

    experience_text = text[experience_match.end():]
    years_match = re.search(
        r"(?<!\d)(\d+)\s*(?:(?:[-–—]|до)\s*(\d+)\s*)?(?:лет|года|год|г\.)",
        experience_text,
    )

    if not years_match:
        return None, None

    first = int(years_match.group(1))
    second = years_match.group(2)
    prefix = experience_text[:years_match.start()].strip()

    if second:
        return first, int(second)
    if prefix.endswith("до"):
        return None, first
    if prefix.endswith("от"):
        return first, None

    return first, first


def normalize_superjob_area(area):
    if pd.isna(area):
        return None

    area = str(area).split(",")[0].split("\n")[0].strip().lower()

    if "удален" in area or "удалён" in area:
        return "удаленно"

    return area
```

```python
experience_bounds = superjob_df["Мета-данные"].apply(superjob_experience_bounds)

vacancies_pt3 = pd.DataFrame({
    "title": superjob_df["Название вакансии"],
    "company": superjob_df["Компания"],
    "salary_from": superjob_df["Зарплата"].apply(superjob_salary_from),
    "salary_to": superjob_df["Зарплата"].apply(superjob_salary_to),
    "salary_currency": superjob_df["Зарплата"].apply(extract_salary_currency),
    "area": superjob_df["Город"].apply(normalize_superjob_area),
    "experience_from": experience_bounds.apply(lambda bounds: bounds[0]),
    "experience_to": experience_bounds.apply(lambda bounds: bounds[1]),
    "skills": superjob_df["Навыки"],
    "desc": superjob_df[["Мета-данные", "Описание"]].fillna("").apply(
        lambda row: "\n".join(value.strip() for value in row if value.strip()),
        axis=1,
    ),
})

vacancies_pt3 = vacancies_pt3.reindex(columns=vacancies_pt1.columns)
```

```python
def extract_city_from_area(area):
    if pd.isna(area):
        return area

    match = re.search(
        r"\bг\.\s*(.*?)(?=,|\n|;|\s+и\s+др\b|$)",
        str(area),
        flags=re.IGNORECASE,
    )

    return match.group(1).strip().lower() if match else area
```
