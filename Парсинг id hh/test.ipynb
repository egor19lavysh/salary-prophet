```python
all_ids = set()

with open("data/data.vacancy_ids.txt", "r") as file:
    for id_ in file.readlines():
        all_ids.add(id_.strip())

with open("data/vacancy_ids.txt", "r") as file:
    for id_ in file.readlines():
        all_ids.add(id_.strip())

with open("data/vacancies.txt", "r") as file:
    for id_ in file.readlines():
        all_ids.add(id_.strip())

with open("data/all_ids.txt", "w") as file:
    for id_ in list(all_ids):
        file.write(id_ + "\n")
```
